```python
import numpy as np
import pandas as pd
```

```python
# Merging 2 dataframe
```

```python
employees = pd.DataFrame({
    'employee_id': [1,2,3,4,5],
    'name' : ['john', 'neha', 'priya', 'bebo', 'lucky'],
    'department' : ['HR', 'IT', 'Finance','HR', 'IT']
})
salaries = pd.DataFrame({
    'employee_id': [1,2,3,6,7],
    'salary' : [60000, 80000, 65000, 70000, 90000],
    'bonus' : [500, 10000, 2000, 5000, 1200 ]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries,on='employee_id',how='inner')
```

```python
pd.merge(employees,salaries,on='employee_id',how='outer')
```

```python
pd.merge(employees,salaries,on='employee_id',how='left')
```

```python
pd.merge(employees,salaries,on='employee_id',how='right')
```

```python
#Concatination od 2 dataframe
```

```python
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B' : ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B' : ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2], axis=1)
```

```python
#Joining of 2 DataFrame
```

```python
df1 = pd.DataFrame({
    'name': ['Alice', 'bob', 'Charlie']
}, index=[1,2,3])

df2 = pd.DataFrame({
    'score': [85,90,75]
}, index= [2,3,4])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2)
```

```python
df1.join(df2,how='outer')
```
