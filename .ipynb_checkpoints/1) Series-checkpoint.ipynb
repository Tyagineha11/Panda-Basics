```python
#SERIES: One-dimensional labeled array capable of holding any data type. 
# the axis labels are collectively called the index  
```

```python
#CREATING SERIES
```

```python
import numpy as np
import pandas as pd
```

```python
labels = ['a', 'b', 'c']
my_lists = [10,20,30]
arr = np.array([10,20,30])
d = {1:10, 2:20, 3:30}
```

```python
pd.Series(my_lists)
```

```python
pd.Series(my_lists, index=labels)
```

```python
pd.Series(arr)
```

```python
pd.Series(d)
```
