```python
import numpy as np
import pandas as pd
```

```python
# Finding MissingData
```

```python
data = {
    'A' : [1, 2, np.nan, 4, 5],
    'B' : [1, 2, 3, 4, 5],
    'C' : [1, 2, 3, np.nan, np.nan],
    'D' : [1 , np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

```python
# Removing Missing Data
```

```python
df
```

```python
df.dropna()
```

```python
df.dropna(thresh= 3)
```

```python
# Filling the missign data
```

```python
df
```

```python
df.fillna(0)
```

```python
df
```

```python
value = {'A' : 0, 'B' : 100, 'C' : 200, 'D' : 300}
df.fillna(value= value)
```

```python
df.fillna(df.mean())
```
